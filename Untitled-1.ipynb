```python
import pandas as pd

url = "https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/insurance.csv"

df = pd.read_csv(url)

new_df = df[["age", "sex", "bmi", "charges"]]

new_df.to_csv("insurance_selected.csv", index=False)

print("Selected columns written successfully.")
```
