```python
import pandas as pd

data = {
    "Name": ["Aravind", "Rahul"],
    "Age": [20, 21],
    "Course": ["Data Science", "AI"]
}

df = pd.DataFrame(data)

# Writing DataFrame to JSON
df.to_json("students.json", orient="records", indent=4)

print("JSON file created successfully")
```

```python
import pandas as pd
import json

json_string = '''
[
    {
        "name":"Kiran",
        "age":22,
        "course":"Machine Learning"
    },
    {
        "name":"Vijay",
        "age":21,
        "course":"Python"
    }
]
'''

data = json.loads(json_string)

df = pd.DataFrame(data)

print(df)
```
